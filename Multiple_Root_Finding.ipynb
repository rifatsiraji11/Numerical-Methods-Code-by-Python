```python
def df(x):
    return 3*x*x - 12*x + 11
```

```python
def newton(x):

    iteration=0

    while True:

        iteration+=1

        xnew=x-f(x)/df(x)

        error=abs(xnew-x)

        if error<0.000001:
            break

        x=xnew

    return xnew,iteration,error
```

```python
starts=[0.5,2.5,4]


for x0 in starts:

    root,it,err=newton(x0)

    print("\nStarting value:",x0)
    print("Root:",round(root,6))
    print("Iterations:",it)
    print("Final Error:",err)
```
