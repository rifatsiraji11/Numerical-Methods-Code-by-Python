```python
def f(x):
    return x**3 - 2*x + 2
```

```python

x = 0
previous = None

print("Iteration\t x")

values = []
```

```python
for i in range(15):

    x_new = x - f(x)/ (3*x*x-2)

    values.append(x_new)

    print(i+1,"\t\t",round(x_new,6))

    if previous != None:

        if abs(x_new-previous)<0.00001:
            print("\nSequence is Converging")
            break

        elif abs(x_new+x)<0.00001:
            print("\nSequence is Oscillating")
            break

    previous=x
    x=x_new
if len(values)==15:
  print("\nSequence is Diverging")
```
